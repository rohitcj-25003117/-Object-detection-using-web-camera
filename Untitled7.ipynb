```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, clear_output

# Load MobileNet-SSD model
net = cv2.dnn.readNet(
    r"C:\Users\CJ ROHIT\mobilenet_iter_73000.caffemodel",
    r"C:\Users\CJ ROHIT\deploy.prototxt"
)

# Object classes
classes = [
    "background", "aeroplane", "bicycle", "bird", "boat",
    "bottle", "bus", "car", "cat", "chair", "cow",
    "diningtable", "dog", "horse", "motorbike", "person",
    "pottedplant", "sheep", "sofa", "train", "tvmonitor"
]

# Start webcam
cap = cv2.VideoCapture(0)

print("Webcam started successfully.")
print("Press Q in the camera window to stop.")

while True:

    ret, frame = cap.read()

    if not ret:
        print("Cannot read webcam.")
        break

    height, width = frame.shape[:2]

    # Create blob
    blob = cv2.dnn.blobFromImage(
        cv2.resize(frame, (300, 300)),
        0.007843,
        (300, 300),
        127.5
    )

    # Object detection
    net.setInput(blob)
    detections = net.forward()

    # Process detections
    for i in range(detections.shape[2]):

        confidence = detections[0, 0, i, 2]

        if confidence > 0.50:

            class_id = int(
                detections[0, 0, i, 1]
            )

            box = detections[0, 0, i, 3:7] * np.array(
                [width, height, width, height]
            )

            x1, y1, x2, y2 = box.astype(int)

            # Object name
            label = classes[class_id]

            text = "{}: {:.2f}%".format(
                label,
                confidence * 100
            )

            # Draw bounding box
            cv2.rectangle(
                frame,
                (x1, y1),
                (x2, y2),
                (0, 255, 0),
                2
            )

            # Display label
            cv2.putText(
                frame,
                text,
                (x1, max(y1 - 10, 20)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 0),
                2
            )

    # Convert BGR to RGB
    frame_rgb = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2RGB
    )

    # Display output inside Jupyter
    clear_output(wait=True)

    plt.figure(figsize=(10, 7))
    plt.imshow(frame_rgb)
    plt.title("Real-Time Object Detection")
    plt.axis("off")
    display(plt.gcf())
    plt.close()

    # Press Q to stop
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()

print("Object detection stopped.")
```

```python
import cv2
import numpy as np

# =========================================================
# WORKSHOP 2 - OBJECT DETECTION USING WEB CAMERA
# =========================================================

# =========================================================
# 1. Load MobileNet-SSD Model
# =========================================================

model_path = r"C:\Users\CJ ROHIT\mobilenet_iter_73000.caffemodel"
config_path = r"C:\Users\CJ ROHIT\deploy.prototxt"

net = cv2.dnn.readNet(
    model_path,
    config_path
)

print("Model loaded successfully.")


# =========================================================
# 2. Object Classes
# =========================================================

classes = [
    "background",
    "aeroplane",
    "bicycle",
    "bird",
    "boat",
    "bottle",
    "bus",
    "car",
    "cat",
    "chair",
    "cow",
    "diningtable",
    "dog",
    "horse",
    "motorbike",
    "person",
    "pottedplant",
    "sheep",
    "sofa",
    "train",
    "tvmonitor"
]


# =========================================================
# 3. Start Laptop Webcam
# =========================================================

cap = cv2.VideoCapture(0)

if not cap.isOpened():

    print("Error: Webcam could not be opened.")

else:

    print("Webcam started successfully.")
    print("Press S = Save detected output")
    print("Press Q = Stop webcam")


    # =====================================================
    # 4. Real-Time Object Detection
    # =====================================================

    while True:

        # Capture frame
        ret, frame = cap.read()

        if not ret:

            print("Error: Cannot read webcam frame.")
            break


        # Get frame dimensions
        height, width = frame.shape[:2]


        # =================================================
        # 5. Create Blob
        # =================================================

        blob = cv2.dnn.blobFromImage(
            cv2.resize(frame, (300, 300)),
            0.007843,
            (300, 300),
            127.5
        )


        # =================================================
        # 6. Object Detection
        # =================================================

        net.setInput(blob)

        detections = net.forward()


        # =================================================
        # 7. Process Detected Objects
        # =================================================

        for i in range(detections.shape[2]):

            confidence = detections[0, 0, i, 2]


            # Confidence threshold
            if confidence > 0.50:

                class_id = int(
                    detections[0, 0, i, 1]
                )


                # =================================================
                # 8. Get Bounding Box
                # =================================================

                box = detections[
                    0, 0, i, 3:7
                ] * np.array(
                    [width, height, width, height]
                )

                x1, y1, x2, y2 = box.astype(int)


                # Keep coordinates inside frame
                x1 = max(0, x1)
                y1 = max(0, y1)
                x2 = min(width - 1, x2)
                y2 = min(height - 1, y2)


                # =================================================
                # 9. Get Object Name
                # =================================================

                label = classes[class_id]


                # =================================================
                # 10. Confidence Percentage
                # =================================================

                text = "{}: {:.2f}%".format(
                    label,
                    confidence * 100
                )


                # =================================================
                # 11. Draw Bounding Box
                # =================================================

                cv2.rectangle(
                    frame,
                    (x1, y1),
                    (x2, y2),
                    (0, 255, 0),
                    2
                )


                # =================================================
                # 12. Display Object Name
                # =================================================

                cv2.putText(
                    frame,
                    text,
                    (x1, max(y1 - 10, 20)),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.6,
                    (0, 255, 0),
                    2
                )


        # =====================================================
        # 13. Display Real-Time Output
        # =====================================================

        cv2.imshow(
            "WORKSHOP 2 - Real-Time Object Detection",
            frame
        )


        # =====================================================
        # 14. Keyboard Controls
        # =====================================================

        key = cv2.waitKey(1) & 0xFF


        # -----------------------------------------------------
        # Press S to Save Output
        # -----------------------------------------------------

        if key == ord("s"):

            output_path = r"C:\Users\CJ ROHIT\workshop2_output.jpg"

            cv2.imwrite(
                output_path,
                frame
            )

            print("==========================================")
            print("Output saved successfully!")
            print("Saved at:")
            print(output_path)
            print("==========================================")


        # -----------------------------------------------------
        # Press Q to Stop
        # -----------------------------------------------------

        if key == ord("q"):

            break


    # =========================================================
    # 15. Release Webcam
    # =========================================================

    cap.release()

    cv2.destroyAllWindows()

    print("Object detection stopped successfully.")
```
